## Упражнение 1. Оценки за ежемесячные тесты

Создадим Series из 10 случайных оценок от 70 до 100 с индексами от сентября до июня.

Найдем:

- среднюю оценку за весь год;
- среднюю за первое полугодие;
- среднюю за второе полугодие;
- изменение успеваемости.

```python
g = np.random.default_rng(0)

months = 'Sep Oct Nov Dec Jan Feb Mar Apr May Jun'.split()

s = pd.Series(
    g.integers(70, 101, 10),
    index=months
)

print("Оценки:")
display(s)

year_average = s.mean()
first_half_average = s['Sep':'Jan'].mean()
second_half_average = s['Feb':'Jun'].mean()

print(f"Средняя за год: {year_average}")
print(f"Средняя за первое полугодие: {first_half_average}")
print(f"Средняя за второе полугодие: {second_half_average}")
print(f"Изменение: {second_half_average - first_half_average}")
```

## Упражнение 2. Масштабирование оценок

Создадим оценки от 40 до 60 и масштабируем их так, чтобы средняя оценка стала равна 80.

```python
g = np.random.default_rng(0)

months = 'Sep Oct Nov Dec Jan Feb Mar Apr May Jun'.split()

s = pd.Series(
    g.integers(40, 60, 10),
    index=months
)

print("Исходные оценки:")
display(s)

print("Исходное среднее:", s.mean())

scaled = s + (80 - s.mean())

print("\nПосле масштабирования:")
display(scaled)

print("Новое среднее:", scaled.mean())
```

## Упражнение 3. Цифра разряда десятков

Создадим 10 случайных чисел от 0 до 99 и получим цифру десятков каждого числа.

Например:

- 47 → 4
- 83 → 8
- 9 → 0

```python
g = np.random.default_rng(0)

s = pd.Series(g.integers(0, 100, 10))

print("Исходные числа:")
display(s)

tens = s // 10

print("Цифры разряда десятков:")
display(tens)
```

## Упражнение 4. Описательная статистика

Создадим 100 000 случайных значений нормального распределения:

- среднее = 0;
- стандартное отклонение = 100.

Посмотрим описательную статистику, затем создадим выброс и сравним результаты.

```python
g = np.random.default_rng(0)

s = pd.Series(
    g.normal(0, 100, 100_000)
)

print("Статистика до изменения:")
display(s.describe())

s.loc[s == s.min()] = 5 * s.max()

print("\nСтатистика после изменения:")
display(s.describe())
```

### Вывод

После появления большого выброса:

- среднее значение изменилось;
- стандартное отклонение увеличилось;
- медиана почти не изменилась.

Это показывает, что медиана менее чувствительна к выбросам, чем среднее значение.

## Упражнение 5. Температура по понедельникам

Создадим данные температуры за 4 недели.

Среднее нормального распределения — 20 °C, стандартное отклонение — 5 °C.

Найдем среднюю температуру по понедельникам.

```python
days = 'Sun Mon Tue Wed Thu Fri Sat'.split()

g = np.random.default_rng(0)

s = pd.Series(
    g.normal(20, 5, 28),
    index=days * 4
).round().astype(np.int8)

print("Температуры:")
display(s)

monday_average = s.loc['Mon'].mean()

print("Средняя температура по понедельникам:", monday_average)
```

## Упражнение 6. Пассажиропоток в такси

Используем данные о количестве пассажиров в поездках на такси.

Найдем долю поездок:

- с одним пассажиром;
- с шестью пассажирами.

```python
passenger_url = (
    "https://gist.githubusercontent.com/reuven/"
    "5d0523fb7a18abcde6f3efcb0bc26f17/raw/"
    "taxi-passenger-count.csv"
)

s = pd.read_csv(
    passenger_url,
    header=None
).squeeze()

print("Первые значения:")
display(s.head())

result = s.value_counts(normalize=True)[[1, 6]]

print("Доля поездок:")
display(result)

print(f"1 пассажир: {result[1] * 100:.2f}%")
print(f"6 пассажиров: {result[6] * 100:.2f}%")
```

## Упражнение 7. Короткие, средние и длинные поездки

Разделим поездки на категории:

- short — до 2 миль включительно;
- medium — больше 2 и до 10 миль включительно;
- long — больше 10 миль.

После этого посчитаем количество поездок каждого типа.

```python
distance_url = (
    "https://gist.githubusercontent.com/reuven/"
    "a0bded849e5925f46ff44686d6f7383e/raw/"
    "taxi-distance.csv"
)

s = pd.read_csv(
    distance_url,
    header=None
).squeeze()

print("Первые дистанции:")
display(s.head())

categories = pd.cut(
    s,
    bins=[0, 2, 10, s.max()],
    include_lowest=True,
    labels=['short', 'medium', 'long']
)

result = categories.value_counts()

print("Количество поездок:")
display(result)
```
